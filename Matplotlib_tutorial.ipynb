```python
import matplotlib.pyplot as plt

#pyplot is a collection of functions that make matplotlib work as MATLAB
```

```python
# 1.CREATE CHART
```

```python
#data
x= [1,2,3,4,5]
y= [10,20,25,35,45]

plt.plot(x,y)               #line chart
plt.show                    #to diaplay chart in jp nb
```

```python
# 2.CUSTOMIZE CHART
```

```python
plt.figure(figsize =(4,3))          #figure size
plt.plot(x, y , color= 'blue', marker='o' , linestyle='dashed' , linewidth= 2, markersize= 10)

plt.title('random graph')
plt.xlabel('hours studied')
plt.ylabel('hours slept')


plt.show()
```

```python
# 3. ADVANCE LINE CHART OR MULTIPLELINES AND LEGENDS
```

```python
#data
x= [1,2,3,4,5]
y1= [10,20,25,35,45]
y2= [20,30,35,45,55]

plt.plot(x, y1, label= 'SALES 2024')
plt.plot(x, y2, label= 'SALES 2025')

plt.title('YEARLY SALES')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.legend()               #TO SHOW LEGEND
plt.show()
```

```python
# 4. BAR CHART
```

```python
x= ['A' , 'B' , 'C', 'D' ,'E']
y= [10,20,25,35,65]

plt.figure(figsize= (4,3))
plt.bar(x,y)
plt.title('BAR CHART EXAMPLE')
plt.show()
```

```python
# 5. HISTOGRAM 
```

```python
import random
data = [random.randint(1,20) for _ in range(500)]

plt.hist(data , bins= 20)        #histogram chart
plt.title('HISTORGAM EXAMPLE')
plt.show()
```

```python
# 6. PIE CHART
```

```python
#USED  TO SHOW PART-TO-WHOLE RELATIONSHIPS/DATA

#data
Categories= ['A' , 'B' , 'C', 'D' ,'E']
Sales= [10,20,25,35,65]

plt.pie(Sales, labels=Categories , autopct =  '%1.1f%%' , startangle=90)
plt.title('PIE CHART EXAMPLE')
plt.show()
```

```python
#7. SCATTER PLOT
```

```python
#USED TO FIND RELATIONSHIP BETWEEN VARIABLES

#data
y1 = [10,20,25,35,45]
y2 =[20,30,35,45,55]

plt.scatter(y1, y2)
plt.title('SCATTER PLOT EXAMPLE')
plt.show()
```

```python
# 8. SUBPLOTS
```

```python
#USED TO SHOW MULTIPLE CHART IN ONE FIGURE
```

```python
#DATA

#data-1  -- bar chart
Categories= ['Mon' , 'Tues' , 'Wed', 'Thurs' ,'Fri']
Sales= [10,20,25,35,65]

#DATA-2 --   scatter plot
y1 = [10,20,25,35,45]
y2 =[20,30,35,45,55]

plt.figure(figsize=(10,4))

#first plot-  bar chart
plt.subplot(1,2,1)      #row , column , position
plt.bar(Categories,Sales)
plt.title('DAILY SALES')
plt.xlabel('Week Days')
plt.ylabel('Sales')




#second plog -- scatter plot
plt.subplot(1,2,2)      #row , column , position
plt.scatter(y1,y2)
plt.title('USER EXAMPLE')
plt.xlabel('User1')
plt.ylabel('User2')

plt.show()
```

```python
#   9.MATPLOTLIB WITH PANDAS   -- REAL DATA
```

```python
#CREATE DATAFRAME 
import pandas as pd

#data 
data =  {
    'Month' : ['Jan' , 'Feb' , 'Mar' , 'Apr'],
    'Sales' : [12000, 11000 , 13000, 30000]    
}

df= pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'] , df['Sales'])
plt.title('Matplotlib with Pandas')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

```python
# 10.SAVE CHARTS/PLOTS
```

```python
plt.bar(df['Month'] , df['Sales'])
plt.title('Matplotlib with Pandas')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.savefig('MONTHLY_SALES_FROM_DF.PNG')

plt.show()
```
